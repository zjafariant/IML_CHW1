## Array creation
>Create the followings to numpy array:
* A: a 3\*3 array like this:
$$
A = \left(\begin{array}{cc} 
10 & 20 & 30\\
5 & 10 & 15\\

$$ 
* B: a 3\*3\*4 that filled with the numbers start from 15 and have steps of 5 to the end. The array should be filled first along the first dimension then the second dimension and at last the third dimension. like this:
$$
B[:, :, 0] = \left(\begin{array}{cc} 
15 & 30 & 45\\
20 & 35 & 50\\
25 & 40 & 55

$$ 

```python
### START CODE HERE ###

A = np.array([[10,20,30],[5,10,15],[1,5,8]])
print(A)

for i in range(4):
    B[:,:,i] = np.arange(15+45*i,60+45*i,5).reshape(3,3).transpose()        
print(B[:,:,0])
print(B)

### END CODE HERE ###
```

## Loading dataset
>Load the IMDB-Movie-Data.csv file into a pandas dataframe and show its first 5 rows. Each row of this dataframe corresponds to one film.

```python
### START CODE HERE ## 
                                       
df = pd.read_csv ('IMDB-Movie-Data.csv')                      #load csv file
df.head(5)                                                    #show the first 5 rows
### END CODE HERE ###
```

## Actors/Actresses Activity
>For each Actors/Actresses caculate the number of movies in the dataframe, then sort the actors/actresses first by their number of movies in descending order and then by their name in ascending order. 

```python
### START CODE HERE ###
actor = []

for i in range(len(df['Actors'])):
    actor += df.loc[i,'Actors'].split(',')                                         # all of name
        
    
actors = list(dict.fromkeys(actor))                                                #names that are not repeated
count = [actor.count(i) for i in actors]                                           #number of repeated names that means number of movie 
d = {'Actors':actors, 'number of movies':count}

df_actors = pd.DataFrame(d)                                                        #creating dataframe of actors and number of movies 
df_actors

display(df_actors.sort_values(by=['number of movies'],ascending=False))
df_actors.sort_values(by=['Actors'],ascending=True)

### END CODE HERE ###
```

## Score Histogram
>Draw the histogram of metascores and ratings. 

```python
### START CODE HERE ###
x = df['Metascore']
y = df['Rating']
fig = plt.figure(figsize=(10,7))                          # creating a figure
plt.subplot(221,xlabel='Metascore')                       # creating subplots
plt.hist(x,bins=10)                                       # histogram of metascore

plt.subplot(222,xlabel='Rating')
plt.hist(y,bins=10)                                       #histogram of rating
### END CODE HERE ###
```

## Correlations
>Calculate the correlation between rating, votes and metascore of the movies.
Is there any meaningful relation between these features? 

```python
### START CODE HERE ###
features1=list(['Rating','Votes','Metascore'])
cor1 = df[features1].corr('pearson')
cor2= df[features1].corr('spearman')
cor3 = df[features1].corr('kendall')
display(cor1,cor2,cor3)
### END CODE HERE ###
```

## File
> Craet a .txt file and name it "Name_actors". You must write the name of all actors/actresses in the IMDB-Movie-Data.csv file. Don't write a name multiple time and write each name in a new line.

```python
### START CODE HERE ###
actors = []
for i in range(len(df['Actors'])):
    actors += df.loc[i,'Actors'].split(',')
        
actors = list(dict.fromkeys(actors))

with open("Name_actors.txt",'a') as f:
    for i in range(len(actors)):
        f.write(actors[i])
        f.write('\n')
        
### END CODE HERE ###
```

## Plot
>You must plot the cos(x) curve. plot 20 cycle of the curve. the frequency of first half of curve is 50 Hz and frequency of second lalf is 200 Hz. 
> set the x_label, y_label and title for your plot.

```python
### START CODE HERE ###

fig = plt.figure(figsize=(15,5))
x = np.arange(0,0.2,0.0005)
y = np.cos(100*np.pi*x)
x2 = np.arange(0.2,0.25,0.0005)
z = np.cos(400*np.pi*x2)
plt.plot(x,y,x2,z)
plt.title('20cycle of curve cos with 50hz and 200hz')
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.show()
### END CODE HERE ###
```
